Since I need to learn patterns of non-linear behaivours I used Random Forest Regressor which will create decision trees across the learning outcomes  then find the average of the outputs and hence results. 

RANDOM FOREST REGRESSOR

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
 # the random forest model, number of tree that are in the forest, depth to avoid overfitting
rf_regressor = RandomForestRegressor(
    n_estimators =100,
    max_depth = 6,
    random_state = 42
    )
```

```python
print("Evaluation")
print(f"Average error{mae_rf} points")            # the average error of during prediction
print(f"R² Score (Model strength): {r2_rf}")   # the inputs are 94% good at predicting the y / 94 % of the difference in thr grades are explained by the learning behaivours
```

I also used the Gradient Boosting Regression model and instead of creating independent trees like the random forest model this model creates sequential  shallow decision trees that correct the mistakes made by the trees before.The model learns from the training X and maps them on the correct answers y then learn patterns. It then creates trees and check for errors then creates a second to correct the  erros and it will do that for n times.

GRADIENT BOOSTING MODEL

```python
from sklearn.ensemble import GradientBoostingRegressor
```

```python
predictions = model.predict(X_test)  # predicting from unseen x-test data by passing them over the 100 trees
```

```python
mae_gb = mean_absolute_error(y_test, predictions)      # mean squared errors checks the error/ distance btwn the actual and prediction values
```

```python
r2_gb = r2_score(y_test, predictions)            # coeeficient of determination
```

TUNING

RANDOM FOREST TUNING

```python
from sklearn.model_selection import GridSearchCV
```

```python
parameter_grid_rf = {                    #parameter grid for rf model
     'n_estimators': [100, 200, 300],
    'max_depth': [4, 6, 8, 10],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}
```

```python
# noe go through the grid search and choose the best combinations and give me the best r2 score
grid_rf = GridSearchCV(         
    RandomForestRegressor(random_state=42),
    parameter_grid_rf,             #this are the parameters that you should go for
    cv=5,
    scoring='r2',
    n_jobs=-1
)
```

```python
# now do the actual tunning with x and y train then try the combinations and get the best combinations
grid_rf.fit(X_train, y_train)         
```

```python
best_rf = grid_rf.best_estimator_  #  get me the best rf model with the best combinations
```

```python
print(f"Best Gb parameters; {best_rf}")
print(f"Best R2 Gb : {grid_rf.best_score_}")       # gives the r2 score from the best combinations
```

Hyperparameter tuning Improved the performance metric of Random Forest model. The MAE in Random forest was improved  from 3.761 to 3.5887 points and the R2 from 0.936 to 0.390.From this the tuned model shows that evaluation metrics have improved from the baseline model.
